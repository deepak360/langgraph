# Prompt Chaining
`when we intract with LLM multiple time in a workflow this is called the prompt chaining`

```python
from langgraph.graph import StateGraph, START, END
from langchain_openai import ChatOpenAI
from typing import TypedDict
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
```

```python
model = ChatOpenAI(api_key=os.getenv("OPENAI_API_KEY"))
```

```python
class BlogState(TypedDict):
    title: str
    outline: str
    content: str
```

```python
def create_outline(state: BlogState) -> BlogState:
    #fetch title
    title = state['title']

    #call llm to generate outline
    prompt = f"Generate an outline for a blog on the topic - {title}"
    outline = model.invoke(prompt).content

    #update state
    state['outline'] = outline

    return state
```

```python
def create_blog(state: BlogState) -> BlogState:
    #fetch title and outline
    title = state['title']
    outline = state['outline']

    prompt = f"Write a detailed blog on the title - {title} using the following outline \n {outline}"

    content = model.invoke(prompt).content

    state['content'] = content

    return state
```

```python
#create graph
graph = StateGraph(BlogState)

#add node
graph.add_node('create_outline', create_outline)
graph.add_node('create_blog', create_blog)

#edges
graph.add_edge(START, 'create_outline')
graph.add_edge('create_outline', 'create_blog')
graph.add_edge('create_blog', END)

#compile
workflow = graph.compile()
```

```python
initial_state = {'title': 'Rise of AI in India'}
final_state = workflow.invoke(initial_state)

print(final_state)
```

```python
#print the outline generated by LLM
print(final_state['outline'])
```

```python
#print the final blog content generated by LLM
print(final_state['content'])
```
